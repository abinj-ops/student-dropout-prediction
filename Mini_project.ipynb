```python
import numpy as np

def transform(df):
    print(f"Original shape: {df.shape}")

    # Transformation 1: Remove duplicate records
    df = df.drop_duplicates()

    # Transformation 2: Handle missing values (Filling numerical with median, categorical with mode)
    # Using median for income/study hours is often safer than mode to avoid outliers
    df['Family_Income'] = df['Family_Income'].fillna(df['Family_Income'].median())
    df['Study_Hours_per_Day'] = df['Study_Hours_per_Day'].fillna(df['Study_Hours_per_Day'].median())
    df['Stress_Index'] = df['Stress_Index'].fillna(df['Stress_Index'].median())
    df['Parental_Education'] = df['Parental_Education'].fillna(df['Parental_Education'].mode()[0])

    # Transformation 3: Change data types (Age and Income make more sense as whole numbers)
    df['Age'] = df['Age'].astype(int)
    df['Family_Income'] = df['Family_Income'].astype(int)

    # Transformation 4: Standardize text (Make Department names uppercase for consistency)
    df['Department'] = df['Department'].str.upper()

    # Transformation 5: Create a calculated column (Total academic commitment time in minutes)
    df['Total_Daily_Commitment_Mins'] = (df['Study_Hours_per_Day'] * 60) + df['Travel_Time_Minutes']

    # Transformation 6: Create categories (Grouping GPA into Performance Categories)
    conditions = [
        (df['CGPA'] >= 3.5),
        (df['CGPA'] >= 2.5) & (df['CGPA'] < 3.5),
        (df['CGPA'] < 2.5)
    ]
    choices = ['Excellent', 'Average', 'Needs Improvement']
    df['Performance_Category'] = np.select(conditions, choices, default='Unknown')

    # Transformation 7: Rename columns (Make them slightly cleaner)
    df = df.rename(columns={'Student_ID': 'ID', 'Study_Hours_per_Day': 'Daily_Study_Hours'})

    # Transformation 8: Filter records (Optional: e.g., removing any impossible age values if they exist, like Age < 10)
    df = df[df['Age'] >= 15]

    print(f"Cleaned shape: {df.shape}")
    print("\nMissing values after transformations:\n", df.isnull().sum())

    return df

raw_data = extract()
cleaned_data = transform(raw_data)
display(cleaned_data.head())
```

```python
# OUTLIER DETECTION USING IQR


numeric_cols = df.select_dtypes(include=np.number).columns

outlier_summary = []

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_summary.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)
```

```python
# Check range of potentially unusual numerical features

check_cols = [
    'Age',
    'Family_Income',
    'Daily_Study_Hours',
    'Attendance_Rate',
    'Assignment_Delay_Days',
    'Travel_Time_Minutes',
    'Stress_Index',
    'Total_Daily_Commitment_Mins'
]

range_check = pd.DataFrame({
    'Minimum': df[check_cols].min(),
    'Maximum': df[check_cols].max(),
    'Mean': df[check_cols].mean(),
    'Median': df[check_cols].median()
})

display(range_check)
```

```python
# ============================================
# OUTLIER TREATMENT USING IQR CAPPING
# ============================================

outlier_treatment_cols = [
    'Family_Income',
    'Daily_Study_Hours'
]

for col in outlier_treatment_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print(f"\n{col}")
    print("Before treatment:")
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)

    # IQR Capping
    df[col] = df[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

    print("After treatment:")
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
#Anything below lower bound → replaced by lower bound
#Anything above upper bound → replaced by upper bound
```

```python
# ============================================================
# FEATURE REDUCTION USING SELECTKBEST
# ============================================================

from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.ensemble import RandomForestClassifier
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Select Top K Features
# ------------------------------------------------------------

# Select top 10 features
k = min(10, X_train.shape[1])

selector = SelectKBest(
    score_func=f_classif,
    k=k
)

# Fit only on training data
X_train_selected = selector.fit_transform(X_train, y_train)

# Apply the same selection to test data
X_test_selected = selector.transform(X_test)

print("Original number of features:", X_train.shape[1])
print("Selected number of features:", X_train_selected.shape[1])


# ------------------------------------------------------------
# 2. Display Selected Features
# ------------------------------------------------------------

# Get feature names
if hasattr(X_scaled, "columns"):
    feature_names = X_scaled.columns.tolist()
else:
    feature_names = [f"Feature_{i}" for i in range(X_train.shape[1])]

selected_features = [
    feature_names[i]
    for i in selector.get_support(indices=True)
]

print("\nSelected Features:")
for i, feature in enumerate(selected_features, 1):
    print(i, ".", feature)


# ------------------------------------------------------------
# 3. Display Feature Scores
# ------------------------------------------------------------

feature_scores = pd.DataFrame({
    "Feature": feature_names,
    "F_Score": selector.scores_,
    "Selected": selector.get_support()
})

feature_scores = feature_scores.sort_values(
    by="F_Score",
    ascending=False
)

print("\n========== FEATURE SCORES ==========")
print(feature_scores)


# ------------------------------------------------------------
# 4. Plot Feature Scores
# ------------------------------------------------------------

top_features = feature_scores.head(k)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"],
    top_features["F_Score"]
)

plt.xlabel("F-Score")
plt.ylabel("Feature")
plt.title("Top Features Selected Using SelectKBest")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 5. Train Random Forest with Selected Features
# ------------------------------------------------------------

rf_selected = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_selected.fit(
    X_train_selected,
    y_train
)

# Prediction
y_pred_selected = rf_selected.predict(X_test_selected)


# ------------------------------------------------------------
# 6. Evaluate Model After Feature Reduction
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_test,
    y_pred_selected
)

precision = precision_score(
    y_test,
    y_pred_selected
)

recall = recall_score(
    y_test,
    y_pred_selected
)

f1 = f1_score(
    y_test,
    y_pred_selected
)

print("\n========== MODEL AFTER FEATURE REDUCTION ==========")

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

```python
# ============================================================
# DECISION THRESHOLD TUNING AFTER SELECTKBEST
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# ------------------------------------------------------------
# 1. Get probability of Dropout = 1
# ------------------------------------------------------------

y_prob = rf_selected.predict_proba(X_test_selected)[:, 1]

print("First 10 predicted probabilities:")
print(y_prob[:10])


# ------------------------------------------------------------
# 2. Test different decision thresholds
# ------------------------------------------------------------

thresholds = np.arange(0.10, 0.91, 0.05)

results = []

for threshold in thresholds:

    y_pred = (y_prob >= threshold).astype(int)

    results.append({
        "Threshold": threshold,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, y_pred, zero_division=0
        )
    })

results = pd.DataFrame(results)

print("\n========== THRESHOLD COMPARISON ==========")
print(results.round(4).to_string(index=False))


# ------------------------------------------------------------
# 3. Find threshold with highest F1 Score
# ------------------------------------------------------------

best_row = results.loc[
    results["F1 Score"].idxmax()
]

best_threshold = best_row["Threshold"]

print("\n========== BEST THRESHOLD ==========")
print("Best Threshold :", best_threshold)
print("Accuracy       :", best_row["Accuracy"])
print("Precision      :", best_row["Precision"])
print("Recall         :", best_row["Recall"])
print("F1 Score       :", best_row["F1 Score"])


# ------------------------------------------------------------
# 4. Final predictions using tuned threshold
# ------------------------------------------------------------

y_pred_threshold = (
    y_prob >= best_threshold
).astype(int)


# ------------------------------------------------------------
# 5. Classification Report
# ------------------------------------------------------------

print("\n========== CLASSIFICATION REPORT ==========")

print(
    classification_report(
        y_test,
        y_pred_threshold,
        zero_division=0
    )
)


# ------------------------------------------------------------
# 6. Confusion Matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    y_test,
    y_pred_threshold
)

print("\n========== CONFUSION MATRIX ==========")
print(cm)

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_threshold
)

plt.title(
    f"Confusion Matrix - Threshold {best_threshold:.2f}"
)

plt.show()


# ------------------------------------------------------------
# 7. F1 Score vs Threshold
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    results["Threshold"],
    results["F1 Score"],
    marker="o"
)

plt.axvline(
    best_threshold,
    linestyle="--",
    label=f"Best Threshold = {best_threshold:.2f}"
)

plt.xlabel("Decision Threshold")
plt.ylabel("F1 Score")
plt.title("F1 Score vs Decision Threshold")
plt.legend()

plt.show()


# ------------------------------------------------------------
# 8. Precision and Recall vs Threshold
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.plot(
    results["Threshold"],
    results["Precision"],
    marker="o",
    label="Precision"
)

plt.plot(
    results["Threshold"],
    results["Recall"],
    marker="o",
    label="Recall"
)

plt.axvline(
    best_threshold,
    linestyle="--",
    label=f"Best Threshold = {best_threshold:.2f}"
)

plt.xlabel("Decision Threshold")
plt.ylabel("Score")
plt.title("Precision and Recall vs Decision Threshold")
plt.legend()

plt.show()
```
